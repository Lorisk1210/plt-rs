Animations in a notebook

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import animation, rc
from IPython.display import HTML
fig, ax = plt.subplots(figsize=(3, 2), dpi=50)
x = np.linspace(0, 2 * np.pi, 30)
line, = ax.plot(x, np.sin(x))
def update(frame):
    line.set_ydata(np.sin(x + frame / 2))
    return [line]
anim = animation.FuncAnimation(fig, update, frames=4, interval=100, blit=True)
plt.close(fig)
```

```python
anim
```

```python
HTML(anim.to_jshtml(default_mode='once'))
```

```python
rc('animation', html='jshtml')
anim
```

```python
if animation.writers.is_available('ffmpeg'):
    rc('animation', html='html5')
    display(HTML(anim.to_html5_video()))
else:
    print('no ffmpeg')
```

```python
anim.save('anim.gif', writer='pillow', fps=5)
from PIL import Image
with Image.open('anim.gif') as gif:
    print(gif.n_frames, gif.size)
```

```python
rc('animation', html='none')
fig, ax = plt.subplots(figsize=(2, 2), dpi=40)
frames = [[ax.imshow(np.full((3, 3), i), vmin=0, vmax=3)] for i in range(3)]
anim2 = animation.ArtistAnimation(fig, frames)
plt.close(fig)
HTML(anim2.to_jshtml(fps=2, embed_frames=True, default_mode='loop'))
```
