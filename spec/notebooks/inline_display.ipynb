`%matplotlib inline`: what shows when, and how

```python
%matplotlib inline
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2 * np.pi, 50)
```

```python
matplotlib.get_backend()
```

```python
plt.plot(x, np.sin(x))
```

```python
plt.plot(x, np.cos(x));
```

```python
plt.plot(x, x)
plt.title('a title')
```

```python
fig, ax = plt.subplots(figsize=(4, 3))
ax.bar(['a', 'b', 'c'], [3, 1, 2])
plt.show()
print('after show')
```

```python
plt.figure(figsize=(3, 2))
plt.plot([1, 3, 2])
plt.figure(figsize=(2, 2), dpi=50)
plt.scatter([1, 2, 3], [3, 1, 2], c=[1, 2, 3])
```

```python
fig, axs = plt.subplots(2, 2, figsize=(4, 4))
axs
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
ax.plot(x, np.tan(x) / 10)
plt.close(fig)
fig
```

```python
ax.set_title('changed in a later cell')
fig
```

```python
from IPython.display import display
fig, ax = plt.subplots(figsize=(3, 2))
ax.plot([0, 1], [0, 1])
display(fig)
ax.set_xlim(0, 2)
display(fig)
plt.close(fig)
```

```python
plt.figure()
plt.plot([1, 2])
plt.close('all')
```

```python
f1 = plt.figure(figsize=(2, 2)); f2 = plt.figure(figsize=(3, 2))
plt.plot([2, 1])
plt.close(f1)
```

```python
plt.gcf()
```

```python
plt.ioff()
fig = plt.figure(figsize=(3, 2))
plt.plot([1, 2, 3])
matplotlib.is_interactive()
```

```python
plt.show()
```

```python
plt.ion()
plt.figure(figsize=(3, 2))
plt.plot([3, 2, 1])
matplotlib.is_interactive()
```

```python
fig = plt.figure(figsize=(3, 2), facecolor='none')
ax = fig.add_subplot()
ax.set_facecolor('none')
ax.plot([1, 2])
```

```python
with plt.rc_context({'xtick.color': 'white', 'ytick.color': 'white'}):
    fig = plt.figure(figsize=(3, 2), facecolor='none')
    fig.add_subplot().plot([1, 2])
```

```python
plt.figure(figsize=(3, 3))
plt.imshow(np.outer(np.sin(x), np.cos(x)), cmap='viridis')
plt.colorbar()
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
print(ax.plot([1, 2], label='line'))
print(ax.legend())
print(ax.get_xaxis(), ax.patch)
fig
```

```python
fig, ax = plt.subplots(figsize=(2, 2))
ax.plot([1, 2])
fig.savefig('figure.png')
import os
print(os.path.getsize('figure.png') > 0)
```
