`%config InlineBackend`: formats, keywords, rc and closing

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 1, 20)
```

```python
%config InlineBackend.figure_formats = ['svg']
plt.figure(figsize=(3, 2)); plt.plot(x, x ** 2);
```

```python
%config InlineBackend.figure_formats = ['pdf']
plt.figure(figsize=(3, 2)); plt.plot(x, x ** 3);
```

```python
%config InlineBackend.figure_formats = ['retina']
plt.figure(figsize=(3, 2)); plt.plot(x, np.sqrt(x));
```

```python
%config InlineBackend.figure_formats = ['jpeg']
plt.figure(figsize=(3, 2)); plt.plot(x, 1 - x);
```

```python
%config InlineBackend.figure_formats = {'png', 'svg'}
plt.figure(figsize=(3, 2)); plt.plot(x, x / 2);
```

```python
%config InlineBackend.figure_format = 'png'
plt.figure(figsize=(3, 2)); plt.plot(x, x / 3);
```

```python
%config InlineBackend.figure_formats = ['gif']
```

```python
%config InlineBackend.print_figure_kwargs = {'bbox_inches': None}
%config InlineBackend.figure_formats = ['png']
plt.figure(figsize=(3, 2)); plt.plot(x, x);
```

```python
%config InlineBackend.print_figure_kwargs = {'bbox_inches': 'tight', 'dpi': 50, 'facecolor': 'lightgray'}
plt.figure(figsize=(3, 2)); plt.plot(x, x);
```

```python
%config InlineBackend.figure_formats = ['jpeg']
%config InlineBackend.print_figure_kwargs = {'bbox_inches': None, 'pil_kwargs': {'quality': 30}}
plt.figure(figsize=(3, 2)); plt.plot(x, x);
```

```python
%config InlineBackend
```

```python
%config InlineBackend.figure_formats = ['png']
%config InlineBackend.print_figure_kwargs = {'bbox_inches': 'tight'}
%config InlineBackend.rc = {'lines.linewidth': 4, 'axes.facecolor': 'lightyellow'}
%matplotlib inline
plt.figure(figsize=(3, 2)); plt.plot(x, x);
print(plt.rcParams['lines.linewidth'])
```

```python
%config InlineBackend.close_figures = False
fig = plt.figure(figsize=(3, 2))
plt.plot(x, x)
```

```python
plt.plot(x, 1 - x)
len(plt.get_fignums())
```

```python
%config InlineBackend.close_figures = True
plt.title('closing again')
```
