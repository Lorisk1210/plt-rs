What people plot inline: projections, layouts, styles, text

```python
%matplotlib inline
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
t = np.linspace(0, 2 * np.pi, 40)
```

```python
fig = plt.figure(figsize=(4, 3))
ax = fig.add_subplot(projection='3d')
ax.plot(np.cos(t), np.sin(t), t)
ax.set_title('3-D')
```

```python
ax = plt.subplot(projection='polar', figure=plt.figure(figsize=(3, 3)))
ax.plot(t, 1 + np.sin(3 * t))
```

```python
plt.figure(figsize=(3, 2))
plt.title(r'$\alpha_i^2 + \frac{\beta}{2}$')
plt.xlabel(r'$\sqrt{x}$')
plt.plot(t, t ** 0.5);
```

```python
fig, axs = plt.subplots(2, 2, figsize=(4, 3), layout='constrained')
for i, ax in enumerate(axs.flat):
    ax.plot(t, np.sin(t + i))
    ax.set_title(f'panel {i}')
fig.suptitle('constrained')
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
ax.plot(t, np.sin(t), label='sin')
ax.plot(t, np.cos(t), label='cos')
ax.legend(loc='center left', bbox_to_anchor=(1, 0.5))
```

```python
fig, ax = plt.subplots(figsize=(3, 2.5))
im = ax.pcolormesh(np.outer(np.sin(t[:10]), np.cos(t[:12])), cmap='coolwarm')
fig.colorbar(im, ax=ax, label='value')
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
ax.plot(t, t, 'r-')
ax.twinx().plot(t, np.exp(t / 3), 'b--')
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
ax.plot(t, np.sin(t))
ax.annotate('peak', xy=(np.pi / 2, 1), xytext=(3, 0.5), arrowprops=dict(arrowstyle='->'))
```

```python
fig = plt.figure(figsize=(4, 2), layout='constrained')
left, right = fig.subfigures(1, 2)
left.subplots().plot([1, 2])
right.subplots(2, 1)
right.suptitle('right')
```

```python
with plt.style.context('ggplot'):
    plt.figure(figsize=(3, 2))
    plt.hist(np.sin(t) * 3, bins=8)
```

```python
matplotlib.rcParams['figure.dpi'] = 50
plt.figure(figsize=(3, 2)); plt.plot([3, 1, 2]);
matplotlib.rcParams['figure.dpi'] = 100
```

```python
plt.figure('named', figsize=(2, 2)); plt.plot([1, 2]);
print(plt.get_figlabels())
```

```python
fig, ax = plt.subplots(figsize=(2, 2))
ax.plot([1, 2])
fig.show()
```

```python
plt.figure(figsize=(2, 2)); plt.plot([2, 1])
plt.show(block=False)
print(plt.get_fignums())
```

```python
from IPython.display import display
fig, ax = plt.subplots(figsize=(2, 2))
ax.plot([1, 2])
display(fig, metadata={'image/png': {'width': 100}})
plt.close(fig)
```

```python
import io
fig, ax = plt.subplots(figsize=(2, 2))
ax.plot([1, 2])
for fmt in ['png', 'svg', 'pdf', 'jpg', 'eps']:
    buf = io.BytesIO()
    fig.savefig(buf, format=fmt)
    print(fmt, buf.getvalue()[:4])
plt.close(fig)
```
