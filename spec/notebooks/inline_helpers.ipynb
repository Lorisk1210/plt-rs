matplotlib-inline's and IPython's helpers

```python
%matplotlib inline
import matplotlib.pyplot as plt
from matplotlib_inline.backend_inline import set_matplotlib_formats, set_matplotlib_close
from IPython.core import pylabtools
```

```python
set_matplotlib_formats('svg')
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
```

```python
set_matplotlib_formats('png', bbox_inches=None, dpi=40)
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
```

```python
set_matplotlib_formats('jpeg', pil_kwargs={'quality': 50})
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
```

```python
set_matplotlib_formats('bmp')
```

```python
set_matplotlib_formats('png')
```

```python
set_matplotlib_close(False)
fig = plt.figure(figsize=(2, 2))
plt.plot([1, 2])
```

```python
plt.plot([2, 1])
plt.get_fignums()
```

```python
set_matplotlib_close(True)
plt.title('last')
```

```python
from IPython.display import set_matplotlib_formats
```

```python
fig, ax = plt.subplots(figsize=(2, 1))
ax.plot([1, 2])
png = pylabtools.print_figure(fig, 'png')
svg = pylabtools.print_figure(fig, 'svg')
print(type(png).__name__, png[:8], type(svg).__name__, svg[:5])
plt.close(fig)
```

```python
print(repr(pylabtools.print_figure(plt.figure(), 'png')))
plt.close('all')
```

```python
fig = plt.figure(figsize=(2, 1))
data, meta = pylabtools.retina_figure(fig)
print(type(data).__name__, meta)
plt.close(fig)
```

```python
a = plt.figure(1); b = plt.figure(2)
print([f.number for f in pylabtools.getfigs()], pylabtools.getfigs(2) == [b])
plt.close('all')
```

```python
pylabtools.figsize(3, 1.5)
print(plt.rcParams['figure.figsize'])
plt.plot([1, 2]);
```
