pandas' and seaborn's plots in a notebook

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
t = np.linspace(0, 1, 30)
df = pd.DataFrame({'a': np.sin(6 * t), 'b': np.cos(6 * t), 'c': t ** 2},
                  index=pd.date_range('2024-01-01', periods=30, freq='D'))
```

```python
df.plot(figsize=(4, 3))
```

```python
df.iloc[:5].plot.bar(figsize=(4, 3));
```

```python
df.hist(figsize=(5, 4));
```

```python
df.plot.scatter(x='a', y='b', c='c', figsize=(4, 3));
```

```python
df['a'].plot.area(figsize=(4, 2), title='area')
```

```python
df.boxplot(figsize=(4, 3))
```

```python
sns.lineplot(data=df, x=df.index, y='a')
```

```python
long = pd.DataFrame({'value': np.concatenate([t, t ** 2, np.sqrt(t)]),
                     'kind': np.repeat(['t', 'square', 'root'], len(t))})
sns.histplot(data=long, x='value', hue='kind', bins=10);
```

```python
plt.figure(figsize=(4, 3))
sns.heatmap(df.iloc[:6].corr(), annot=True, fmt='.2f');
```

```python
sns.relplot(data=long, x='value', y=np.arange(len(long)), col='kind', height=2)
```

```python
sns.set_theme()
sns.scatterplot(x=t, y=t ** 3);
```

```python
sns.reset_orig()
plt.plot(t, t);
```
