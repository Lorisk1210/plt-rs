`%matplotlib` switching between backends

```python
import matplotlib
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
matplotlib.get_backend()
```

```python
%matplotlib widget
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
matplotlib.get_backend()
```

```python
%matplotlib notebook
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
matplotlib.get_backend()
```

```python
%matplotlib inline
plt.close('all')
plt.figure(figsize=(2, 2)); plt.plot([2, 1]);
matplotlib.get_backend()
```

```python
%matplotlib agg
plt.figure(figsize=(2, 2)); plt.plot([1, 2])
plt.show()
matplotlib.get_backend()
```

```python
plt.close('all')
matplotlib.use('svg')
plt.figure(figsize=(2, 2)); plt.plot([1, 2])
matplotlib.get_backend()
```

```python
%matplotlib module://matplotlib_inline.backend_inline
plt.figure(figsize=(2, 2)); plt.plot([1, 2]);
matplotlib.get_backend()
```
