`%matplotlib widget`: ipympl's canvas, driven through its comm

```python
%matplotlib widget
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
events = []
```

```python
matplotlib.get_backend()
```

```python
fig, ax = plt.subplots(figsize=(4, 3), dpi=100)
ax.plot(np.linspace(0, 10, 11), np.linspace(0, 10, 11) ** 0.5)
ax.set_xlim(0, 10)
ax.set_ylim(0, 4)
def record(event):
    xdata, ydata = getattr(event, 'xdata', None), getattr(event, 'ydata', None)
    events.append((event.name, str(getattr(event, 'button', None)), getattr(event, 'key', None),
                   xdata if xdata is None else round(xdata, 3),
                   ydata if ydata is None else round(ydata, 3)))
for name in ['button_press_event', 'button_release_event', 'key_press_event',
             'scroll_event', 'resize_event']:
    fig.canvas.mpl_connect(name, record)
```

```python
def fail(event):
    raise ValueError('a failing callback')
cid = fig.canvas.mpl_connect('button_press_event', fail)
```

```python
fig.canvas.mpl_disconnect(cid)
```
