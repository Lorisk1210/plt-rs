matplotlib.widgets in ipympl's canvas

```python
%matplotlib widget
import matplotlib.pyplot as plt
from matplotlib.widgets import Button, CheckButtons, Slider, RadioButtons
changes = []
```

```python
fig = plt.figure(figsize=(4, 3), dpi=100)
slider = Slider(fig.add_axes((0.25, 0.8, 0.5, 0.1)), 'value', 0, 10, valinit=5)
button = Button(fig.add_axes((0.1, 0.5, 0.3, 0.15)), 'press')
check = CheckButtons(fig.add_axes((0.55, 0.4, 0.4, 0.3)), ['one', 'two'], [True, False])
radio = RadioButtons(fig.add_axes((0.1, 0.05, 0.3, 0.3)), ['red', 'blue'])
slider.on_changed(lambda value: changes.append(('slider', round(value, 3))))
button.on_clicked(lambda event: changes.append(('button',)))
check.on_clicked(lambda label: changes.append(('check', label, check.get_status())))
radio.on_clicked(lambda label: changes.append(('radio', label)))
```
